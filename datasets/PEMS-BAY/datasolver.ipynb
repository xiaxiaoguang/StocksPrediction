```python
import pickle
import torch
import numpy as np


def load_pkl(pickle_file: str) -> object:
    """Load pickle data.

    Args:
        pickle_file (str): file path

    Returns:
        object: loaded objected
        
    """

    try:
        with open(pickle_file, "rb") as f:
            pickle_data = pickle.load(f)
    except UnicodeDecodeError:
        with open(pickle_file, "rb") as f:
            pickle_data = pickle.load(f, encoding="latin1")
    except Exception as e:
        print("Unable to load data ", pickle_file, ":", e)
        raise
    return pickle_data

data_file_path = "data_in2016_out12.pkl"

data = load_pkl(data_file_path)
data = data['processed_data']
print(data.shape)
```

```python
tmpl=12
outl=12
```

```python
file_name = f'data_in{tmpl}_out{outl}.pkl'
data = {'processed_data':data}
with open(file_name, 'wb') as file:
    pickle.dump(data, file)
```

```python
length = data['processed_data'].shape[0]-tmpl-outl
testl = int(length*0.2)
validl = int(length*0.1)
trainl = int(length)-testl-validl
rl = 1
tl = 1

idx = {}
train = []
for i in range(0,trainl,rl):
    train.append((i,i+tmpl,i+tmpl+outl))
idx['train']=train
valid=[]
for i in range(trainl,trainl+validl,tl):
    valid.append((i,i+tmpl,i+tmpl+outl))
idx['valid']=valid
test=[]
for i in range(trainl+validl,trainl+testl+validl,tl):
    test.append((i,i+tmpl,i+tmpl+outl))
idx['test'] =test

file_name = f'index_in{tmpl}_out{outl}.pkl'

with open(file_name,'wb') as file:
    pickle.dump(idx,file)

print(train)
print(len(idx['train']),len(idx['test']),len(idx['valid']))
```
