```python
import sys
sys.path.append('/home/wbyan/stock_data')
import numpy as np
import pickle
from data_utils import load_data #会显示报错，但是上面加了sys path可以使用
#读取股市数据
date_list, features, labels, stock_list = load_data(
        data_type="ch_daily_processed",
        stock_list="csi300",
        start_date="2010-01-01",
        end_date="2023-09-30",
        contain_bj=False,
        drop_short=False,
        short_len=0,
)
```

```python
print(features[2,700])
print(features.shape)
features = features.permute(1,0,2)
print(features[700,2])
print(features.shape)
features=features.numpy()
```

```python
file_name = 'data_in144_out12.pkl'
data = {'processed_data':features}
with open(file_name, 'wb') as file:
    pickle.dump(data, file)
print(data)
```

```python
length = features.shape[0]
trainl = int(length*0.7)
testl = int(length*0.15)
validl = int(length-trainl-testl)
print(trainl,testl,validl)

idx = {}
train = []
for i in range(trainl):
    train.append((i,i+144,i+156))
idx['train']=train
valid=[]
for i in range(trainl,trainl+testl):
    valid.append((i,i+144,i+156))
idx['valid']=valid
test=[]
for i in range(trainl+testl,length-156):
    test.append((i,i+144,i+156))
idx['test'] =test

file_name = 'index_in144_out12.pkl'

with open(file_name,'wb') as file:
    pickle.dump(idx,file)

print(idx['test'])
```

```python
scaler = {}
scaler['func']='re_standard_transform'
scaler['args']={'mean': 0, 'std': 1}

file_name = 'scaler_in144_out12.pkl'

with open(file_name,'wb') as file:
    pickle.dump(scaler,file)
```
