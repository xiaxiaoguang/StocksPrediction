```python
import pickle

import torch
import numpy as np


def load_pkl(pickle_file: str) -> object:
    """Load pickle data.

    Args:
        pickle_file (str): file path

    Returns:
        object: loaded objected
        
    """

    try:
        with open(pickle_file, "rb") as f:
            pickle_data = pickle.load(f)
    except UnicodeDecodeError:
        with open(pickle_file, "rb") as f:
            pickle_data = pickle.load(f, encoding="latin1")
    except Exception as e:
        print("Unable to load data ", pickle_file, ":", e)
        raise
    return pickle_data

data_file_path = "scaler_in2016_out12.pkl"
data_file_path2 = "data_in12_out12.pkl"

data = load_pkl(data_file_path)
data2 = load_pkl(data_file_path2)

print(data)
# print(data2)
# print(list(data['train'][2]))
# print(type(data))
# print(len(data))
```
