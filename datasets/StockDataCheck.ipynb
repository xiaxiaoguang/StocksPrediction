Open: The opening price of the stock for the given time period.
    
Pct_chg: The percentage change in the stock price from the previous day's close.

High: The highest price reached during the time period.

Low: The lowest price reached during the time period.

Close: The closing price of the stock for the given time period.

Pre_close: The closing price of the stock for the previous time period.

Ma5, Ma10, Ma15, Ma20, Ma25: Moving averages over different periods, representing smoothed trends in the stock 
price.

Change: The absolute change in the stock price from the previous day's close.

Vol: Trading volume (number of shares traded).

Amount: The total value of all shares traded.

Turnover_rate: The ratio of the total shares traded to the total shares outstanding.

Turnover_rate_f: The ratio of the total shares traded to the free-float shares outstanding.

Volume_ratio: The ratio of the trading volume to the average trading volume over a specific period.

PE, PE_ttm, PB, PS, PS_ttm: Valuation ratios such as Price-to-Earnings, Price-to-Book, and Price-to-Sales.

Total_mv, Circ_mv: Total market value and circulating market value of the company's outstanding shares.

Buy_sm_vol, Buy_sm_amount, Sell_sm_vol, Sell_sm_amount, ...: Trading volumes and amounts for small, medium, large, 
and extra-large investors, both in buying and selling categories.

Net_mf_vol, Net_mf_amount: Net trading volumes and amounts by mutual funds.

Up_limit, Down_limit: The percentage limit up or down to which a stock can move in a single trading session.

Industry: The industry to which the stock belongs.
