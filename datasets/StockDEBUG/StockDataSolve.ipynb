```python

in_length=96
out_length=5
length = features.shape[0]-in_length-out_length
trainl = int(length*0.7)
testl = int(length*0.15)
validl = int(length-trainl-testl)
print(trainl,testl,validl)

idx = {}
train = []
for i in range(trainl):
    train.append((i,i+in_length,i+in_length+out_length))
idx['train']=train
valid=[]
for i in range(trainl,trainl+validl):
    valid.append((i,i+in_length,i+in_length+out_length))
idx['valid']=valid
test=[]
for i in range(trainl+validl,trainl+testl+validl):
    test.append((i,i+in_length,i+in_length+out_length))
idx['test'] =test

file_name = f'index_in{in_length}_out{out_length}.pkl'

with open(file_name,'wb') as file:
    pickle.dump(idx,file)

print(idx['test'],trainl+testl)
```
